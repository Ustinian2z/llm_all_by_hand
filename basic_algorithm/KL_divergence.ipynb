### 交叉熵

```python
import numpy as np

def CrossEntropy(y, y_hat):
    eps = 1e-8
    y_hat = np.clip(y_hat, eps, 1-eps)
    
    if y.ndim == 1 or y.shape[1] == 1:
        return -np.sum(y * np.log(y_hat) + (1-y) * np.log(1 - y_hat))
    
    return -np.mean(np.sum(y * np.log(y_hat), axis = 1))
```

```python
y_true = np.array([[0, 1, 0], [1, 0, 0], [0, 0, 1], [0, 1, 0], [1, 0, 0]])
y_pred = np.array([[0.1, 0.8, 0.1], [0.9, 0.05, 0.05], [0.1, 0.1, 0.8], [0.3, 0.6, 0.1], [0.8, 0.1, 0.1]])

print(CrossEntropy(y_true, y_pred))
```

### KL 散度

```python
import numpy as np

def KLdivergence(y, y_hat):
    eps = 1e-8
    y = np.clip(y, eps, 1-eps)
    y_hat = np.clip(y_hat, eps, 1-eps)

    if y.ndim == 1 or y.shape[1] == 1:
        return np.sum(y * np.log(y/y_hat) + (1-y) * np.log((1-y)/(1-y_hat)))
    
    return np.mean(np.sum(y * np.log(y/y_hat), axis = 1))

y_true = np.array([[0, 1, 0], [1, 0, 0], [0, 0, 1], [0, 1, 0], [1, 0, 0]])
y_pred = np.array([[0.1, 0.8, 0.1], [0.9, 0.05, 0.05], [0.1, 0.1, 0.8], [0.3, 0.6, 0.1], [0.8, 0.1, 0.1]])

print(KLdivergence(y_true, y_pred))
```
