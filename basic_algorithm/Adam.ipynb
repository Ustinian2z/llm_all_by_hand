```python
import numpy as np

class Adam:
    def __init__(self, params, lr=1e-5, beta1=0.9, beta2=0.99, eps=1e-8):
        self.params = params
        self.lr = lr
        self.beta1 = beta1
        self.beta2 = beta2
        self.eps = eps
        self.m = [np.zeros_like(p) for p in params]
        self.v = [np.zeros_like(p) for p in params]
        self.t = 0
    
    def step(self, grads):
        self.t += 1
        for i, (param, grad) in enumerate(zip(self.params, grads)):
            self.m[i] = self.beta1 * self.m[i] + (1-self.beta1) * grad
            self.v[i] = self.beta2 * self.v[i] + (1-self.beta2) * (grad**2)
            m_hat = self.m[i] / (1-self.beta1**self.t)
            v_hat = self.v[i] / (1-self.beta2**self.t)
            
            update = self.lr * m_hat / (np.sqrt(v_hat) + self.eps)
            self.params[i][:] -= update  # 用[:]确保修改原数组，而非创建新数组
        return self.params
```

```python
import numpy as np
def test_adam_optimizer():
    # 1. 初始化待优化的参数（x=2, y=3，目标是优化到0附近）
    params = [np.array([2.0, 4.0, 7.0]), np.array([3.0, 6.0, 9.0])]  # x和y，用数组是为了保持和原代码的维度一致性
    print(f"初始参数: x={params[0]}, y={params[1]}")
    
    # 2. 初始化Adam优化器（调整学习率到0.1，收敛更快）
    optimizer = Adam(params, lr=0.1)
    
    # 3. 迭代优化100次
    for epoch in range(100):
        # 计算目标函数 f(x,y) = x² + y² 的梯度
        x, y = params[0], params[1]
        grad_x = 2 * x  # df/dx = 2x
        grad_y = 2 * y  # df/dy = 2y
        grads = [grad_x, grad_y]
        
        # 执行一次Adam更新
        optimizer.step(grads)
        
        # 每20次迭代打印一次参数，观察变化
        if (epoch + 1) % 20 == 0:
            print(f"迭代 {epoch+1} 次后: x={params[0]}, y={params[1]}")
    
    # 4. 打印最终结果
    print(f"\n优化完成后参数: x={params[0]}, y={params[1]}")
    print(f"目标函数值 f(x,y) = {params[0]**2 + params[1]**2}")

# 运行测试
if __name__ == "__main__":
    test_adam_optimizer()
```
